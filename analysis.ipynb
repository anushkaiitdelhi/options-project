```python
import numpy as np
import matplotlib.pyplot as plt
from pricing_models import black_scholes_call

print("Modules loaded successfully.")
```

```python
# Define parameters
S, K, T, r = 100, 100, 1.0, 0.05
volatilities = np.linspace(0.1, 0.5, 10) # Testing volatility from 10% to 50%

# Calculate prices for each volatility
prices = [black_scholes_call(S, K, T, r, v) for v in volatilities]

print("Calculation complete.")
```

```python
plt.figure(figsize=(8, 5))
plt.plot(volatilities, prices, marker='o', linestyle='-', color='blue')
plt.title('Impact of Volatility on Call Option Price')
plt.xlabel('Volatility ($\sigma$)')
plt.ylabel('Option Price ($)')
plt.grid(True)
plt.show()
```
